```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Social_media_impact_on_life.csv") # load
```

```python
df
```

### Numarical vs Numarical cols_

```python
df = df[~df['Gender'].isin(['Non-Binary', 'Prefer not to say'])]
```

```python
plt.figure(figsize=(5,4))
sns.scatterplot(x="Age", y="Daily_Usage_Hours", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df[["Age", "Daily_Usage_Hours"]].corr()
plt.tight_layout()

plt.figure(figsize=(5,4))
sns.scatterplot(x="Age", y="Weekend_Extra_Hours", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df[["Age", "Weekend_Extra_Hours"]].corr()
plt.tight_layout()

plt.figure(figsize=(5,4))
sns.scatterplot(x="Age", y="Sleep_Duration_Hours", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1),loc='upper left')
df[["Age", "Sleep_Duration_Hours"]].corr()
plt.tight_layout()

plt.figure(figsize=(5,4))
sns.scatterplot(x="Age", y="Perceived_Stress_Score", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1),loc='upper left')
df[["Age", "Perceived_Stress_Score"]].corr()
plt.tight_layout()

plt.figure(figsize=(5,4))
sns.scatterplot(x="Age", y="Mental_Health_Index", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1),loc='upper left')
df[["Age", "Mental_Health_Index"]].corr()
plt.tight_layout()

plt.figure(figsize=(5,4))
sns.scatterplot(x="Age", y="Academic_Performance_GPA", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1),loc='upper left')
df[["Age", "Academic_Performance_GPA"]].corr()
plt.tight_layout()
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x="Daily_Usage_Hours", y="Sleep_Duration_Hours", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.04,1))
df[["Daily_Usage_Hours", "Sleep_Duration_Hours"]].corr()

plt.figure(figsize=(5,3))
sns.scatterplot(x="Daily_Usage_Hours", y="Sleep_Quality_Score", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.04,1))
df[["Daily_Usage_Hours", "Sleep_Quality_Score"]].corr()

plt.figure(figsize=(5,3))
sns.scatterplot(x="Daily_Usage_Hours", y="Perceived_Stress_Score", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.04,1))
df[["Daily_Usage_Hours", "Perceived_Stress_Score"]].corr()

plt.figure(figsize=(5,3))
sns.scatterplot(x="Daily_Usage_Hours", y="Mental_Health_Index", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.04,1))
df[["Daily_Usage_Hours", "Mental_Health_Index"]].corr()

plt.figure(figsize=(5,3))
sns.scatterplot(x="Daily_Usage_Hours", y="Mental_Health_Index", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.04,1))
df[["Daily_Usage_Hours", "Mental_Health_Index"]].corr()
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x="Sleep_Duration_Hours", y="Perceived_Stress_Score", data=df,alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df[["Sleep_Duration_Hours", "Perceived_Stress_Score"]].corr()

plt.figure(figsize=(5,3))
sns.scatterplot(x="Sleep_Duration_Hours", y="Sleep_Quality_Score", data=df,alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df[["Sleep_Duration_Hours", "Sleep_Quality_Score"]].corr()

plt.figure(figsize=(5,3))
sns.scatterplot(x="Sleep_Duration_Hours", y="Mental_Health_Index", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df[["Sleep_Duration_Hours", "Mental_Health_Index"]].corr()
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x="Mental_Health_Index", y="Perceived_Stress_Score", data=df, alpha=0.6, hue='Gender')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df[["Mental_Health_Index", "Perceived_Stress_Score"]].corr()
```

### Numeric vs Categorical cols_

#### Gender

```python
# Keep rows where Gender is NOT in the specified list
df = df[~df['Gender'].isin(['Non-Binary', 'Prefer not to say'])]
```

```python
plt.figure(figsize=(5,3))
sns.boxplot(x="Gender", y="Daily_Usage_Hours", data=df, hue='Academic_Level')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df.groupby("Gender")["Daily_Usage_Hours"].mean()

plt.figure(figsize=(5,3))
sns.boxplot(x="Gender", y="Weekend_Extra_Hours", data=df, hue='Academic_Level')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df.groupby("Gender")["Weekend_Extra_Hours"].mean()

plt.figure(figsize=(5,3))
sns.boxplot(x="Gender", y="Sleep_Duration_Hours", data=df, hue='Academic_Level')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df.groupby("Gender")["Sleep_Duration_Hours"].mean()

plt.figure(figsize=(5,3))
sns.boxplot(x="Gender", y="Sleep_Quality_Score", data=df, hue='Academic_Level')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df.groupby("Gender")["Sleep_Quality_Score"].mean()

plt.figure(figsize=(5,3))
sns.boxplot(x="Gender", y="Perceived_Stress_Score", data=df, hue='Academic_Level')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df.groupby("Gender")["Perceived_Stress_Score"].mean()

plt.figure(figsize=(5,3))
sns.boxplot(x="Gender", y="Mental_Health_Index", data=df, hue='Academic_Level')
plt.legend(title='chart', bbox_to_anchor=(1.05,1))
df.groupby("Gender")["Mental_Health_Index"].mean()
```
